```python
# Load  Dataset
import pandas as pd

df= pd.read_csv("Telco_Customer_Churn_Dataset  (3).csv")
```

```python
# Check Dataste Information (Number of Rows, Number of Columns, & Data Type)
df.info()
```

```python
# Statistical  SUmmary

df.describe()
```

<h3>Data Cleaning<h3>
    

```python
# Check Duplicate Value in Our Dataset

df.duplicated().sum()
```

```python
# In this dataset Duplicate value is 0 so, we cannot need to remove Duplicate value
```

```python
# Convert TotalCharges to Number

df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
```

```python
# Handle Missing Values

df.isnull().sum()
```

```python
# There have TotalCharges is 11 Now Handle missing Values.  Fill Missing Values

df.fillna({"TotalCharges": df["TotalCharges"].median()}, inplace=True)
```

```python
# Verify the missing value is fill.

df["TotalCharges"].isnull().sum()
```

```python
# Standardize Columns Names

df.columns = df.columns.str.lower()
```

<h4> Exploratory Data Analysis (EDA) <h4>

```python
# Summary Statistics

df.describe()
```

<h3> Histogram<h3>

```python
import matplotlib.pyplot as plt


plt.hist(df["monthlycharges"],bins= 20)
plt.title ("Monthlycharges")
plt.show()
```

```python
# Box Plot

plt.boxplot(df["monthlycharges"])
plt.show()
```

```python
#  Churm count


df["churn"].value_counts()
```

```python
# Churn percentage 

(df["churn"].value_counts(normalize=True)*100).round(2)
```

# Customer Segmentation


```python
# Makes Tenure Groups

df["tenure_group"] = pd.cut (
    df["tenure"],
       bins = [0,12,36,72],
       # labels=["0-12","13-36","37+"]
)
```

```python
# Pie Chart

df["tenure_group"].value_counts().plot(
    kind="pie",labels=None,colors=["lavender","magenta","midnightblue"],autopct="%1.1f%%")
```

```python
# Average Monthly Changes

df.groupby("tenure_group")["monthlycharges"].mean().plot(kind="bar")
```

# Advanced Analysis


```python
# Churn by Gender

pd.crosstab(df["gender"],df["churn"])
```

```python
#  Churn by Contract

pd.crosstab(df["contract"],df["churn"])
```

```python
#  Churn by Senior Citizen

pd.crosstab(df["seniorcitizen"],df["churn"])
```

```python
#  Churn by Payment Method


pd.crosstab(df["paymentmethod"],df["churn"])
```

```python
# Tenure Group Analysis


(df.groupby("tenure_group")[["monthlycharges","totalcharges"]].agg(["mean","median","count"])).round(2)
```

```python
#   Churn Rate by Tenure Group

(pd.crosstab(df["tenure_group"],df["churn"], normalize= "index")*100).round(2)
```
